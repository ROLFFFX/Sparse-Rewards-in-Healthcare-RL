```python
import pandas as pd
import pickle
import os
project_dir = os.path.expanduser('../')
```

```python
df_train_shifted = pd.read_csv(
    project_dir + 'data/traj_shifted_train.csv',
    dtype={"a:action": "Int64", "a:next_action": "Int64"}
)
SA_mask = pd.read_pickle(project_dir + "data/behavior_policy/SA_mask_shifted.pkl")
with open(project_dir + 'data/env_model/gymP_shifted_plus.pkl', 'rb') as f:
    gymP_plus = pickle.load(f)
with open(project_dir + 'data/env_model/gymP_shifted_minus.pkl', 'rb') as f:
    gymP_minus = pickle.load(f)
with open(project_dir + 'data/env_model/gymP_shifted_mixed.pkl', 'rb') as f:
    gymP_mixed = pickle.load(f)
```

```python
def trace_state_trajectories(df, target_state, max_print=3):
    # 1. find all trajectories that reach the target_state
    reached = df[df["s:next_state"] == target_state]["traj"].unique()
    print(f"State {target_state} was reached in {len(reached)} trajectories: {reached[:10]}")
    
    if len(reached) == 0:
        print("No trajectories reach this state.")
        return

    # 2. for each trajectory that reaches it
    for i, t_id in enumerate(reached[:max_print]):
        traj_df = df[df["traj"] == t_id].sort_values("step")
        idx = traj_df.index[traj_df["s:next_state"] == target_state][0]
        entry_step = traj_df.loc[idx, "step"]
        before = traj_df[traj_df["step"] <= entry_step]
        after = traj_df[traj_df["step"] > entry_step]
        print(f"\n=== Trajectory {t_id} ===")
        print(f" Reached state {target_state} at step {entry_step}")
        print("\n Context leading up to it:")
        print(before.tail(5)[["step", "s:state", "a:action", "s:next_state", "r:reward"]])
        
        if len(after) == 0:
            print("\n No recorded transitions after reaching this state.")
        else:
            print("\n Future transitions after reaching it:")
            print(after.head(10)[["step", "s:state", "a:action", "s:next_state", "r:reward"]])
```

```python
trace_state_trajectories(df_train_shifted,target_state=330)
```
